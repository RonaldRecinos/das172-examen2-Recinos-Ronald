```python
%%writefile aero_matrix.py
def validar_matrices(cargas, capacidades):
    N = len(cargas)
    if N < 2: return False
    M = len(cargas[0])
    if M < 2: return False
    if len(capacidades) != N: return False

    for i in range(N):
        if len(cargas[i]) != M or len(capacidades[i]) != M: return False
        for j in range(M):
            if cargas[i][j] < 0 or capacidades[i][j] <= 0: return False
    return True

def calcular_ocupacion(cargas, capacidades):
    porcentajes = []
    sobrecargadas = []
    for i in range(len(cargas)):
        fila_porc = []
        for j in range(len(cargas[0])):
            ocupacion = (cargas[i][j] / capacidades[i][j]) * 100.0
            fila_porc.append(ocupacion)
            if ocupacion > 100.0:
                sobrecargadas.append((i, j))
        porcentajes.append(fila_porc)
    return porcentajes, sobrecargadas

def evaluar_balance(cargas, tolerancia_kg):
    N, M = len(cargas), len(cargas[0])
    pesos_filas = [sum(fila) for fila in cargas]
    suma_izq = sum(cargas[i][j] for i in range(N) for j in range(M // 2))
    inicio_der = M // 2 if M % 2 == 0 else (M // 2) + 1
    suma_der = sum(cargas[i][j] for i in range(N) for j in range(inicio_der, M))
    desbalance = abs(suma_izq - suma_der)
    return pesos_filas, desbalance, desbalance <= tolerancia_kg

def extraer_submatriz_critica(porcentajes, k, p):
    max_promedio = -1
    submatriz_critica = []
    for i in range(len(porcentajes) - k + 1):
        for j in range(len(porcentajes[0]) - p + 1):
            suma_sub = sum(porcentajes[i+x][j+y] for x in range(k) for y in range(p))
            promedio = suma_sub / (k * p)
            if promedio > max_promedio:
                max_promedio = promedio
                submatriz_critica = [[porcentajes[i+x][j+y] for y in range(p)] for x in range(k)]
    return submatriz_critica

%%writefile main.py
import aero_matrix as am

cargas = [[500, 100, 480], [600, 200, 650], [300, 50, 300]]
capacidades = [[600, 600, 600], [600, 600, 600], [400, 400, 400]]

if am.validar_matrices(cargas, capacidades):
    porc, sobre = am.calcular_ocupacion(cargas, capacidades)
    pesos, desbal, estado = am.evaluar_balance(cargas, 100)
    sub = am.extraer_submatriz_critica(porc, 2, 2)

    print(f"Celdas sobrecargadas: {sobre}")
    print(f"Desbalance: {desbal}kg | Aprobado: {estado}")
    print(f"Submatriz crítica (2x2): {sub}")
else:
    print("Error dimensional en matrices.")

%%writefile test_aero.py
import unittest
import aero_matrix as am

class TestAeroMatrix(unittest.TestCase):
    def test_validar_matrices(self):
        cargas = [[10, 10], [10, 10]]
        capas = [[20, 20], [20, 20]]
        self.assertTrue(am.validar_matrices(cargas, capas))
        self.assertFalse(am.validar_matrices([[10]], [[20]])) # Falla por tamaño < 2x2

if __name__ == '__main__':
    unittest.main()

%%writefile .gitignore
__pycache__/
*.pyc

# Commit 1
!git add aero_matrix.py
!git commit -m "feat: módulo principal de cálculos matriciales puros"

# Commit 2
!git add main.py
!git commit -m "feat: script de ejecución con casos de prueba"

# Commit 3
!git add test_aero.py
!git commit -m "test: pruebas unitarias de validación dimensional"

# Commit 4
!git add .gitignore
!git commit -m "chore: exclusión de caché de python"

# Subir a la web
!git push origin main
```
